```python
#Convert string dates to actual dates
DNS_df['Timestamp'] = pd.to_datetime(DNS_df['Timestamp'])
NTP_df['Timestamp'] = pd.to_datetime(NTP_df['Timestamp'])

#Set time periods we will record
attack_start = pd.Timestamp('2018-12-01 10:52:00')
attack_end   = pd.Timestamp('2018-12-01 11:05:00')
normal_start = pd.Timestamp('2018-12-01 10:34:00')

#filter dataset to begin at normal start
DNS_df_filtered = DNS_df[(DNS_df['Timestamp'] >= normal_start) & (DNS_df['Timestamp'] <= attack_end)]
NTP_df_filtered = NTP_df[(NTP_df['Timestamp'] <= normal_start) & (NTP_df['Label'] == 'BENIGN')]
# This shows us how many normal vs attack rows we kept
print(DNS_df_filtered['Label'].value_counts())
print(NTP_df_filtered['Label'].value_counts())

# This shows how many rows we kept vs the original 5 million
print(f"Rows kept DNS: {len(DNS_df_filtered)} out of {len(DNS_df)}")
print(f"Rows kept NTP: {len(NTP_df_filtered)} out of {len(NTP_df)}")
```

```python
combined_df = pd.concat([NTP_df_filtered, DNS_df_filtered], ignore_index=True)
combined_df = combined_df.sort_values('Timestamp').reset_index(drop=True)

print(combined_df['Label'].value_counts())
print(f"Total rows: {len(combined_df)}")
```

```python
#Create function to create windows
def create_windows(df, window_size_seconds):
    df = df.copy()
    
    #Round each timestamp down to the nearest window e.g (Windows 5s) Timestamp = 10:51:43.812 -> 10:51:40
    df['window'] = df['Timestamp'].dt.floor(f'{window_size_seconds}s')

    #Splits the dataframe into groups, one per time window
    grouped = df.groupby('window')

    #Total packets (fwd + backward) in the window
    packets_per_window = grouped.apply(lambda x: (x['Total Fwd Packets'] + x['Total Backward Packets']).sum(), include_groups=False)
    #Total bytes transferred in the window
    bytes_per_window = grouped.apply(lambda x: (x['Total Length of Fwd Packets'] + x['Total Length of Bwd Packets']).sum(), include_groups=False)
    #The % of rows using the UDP protocol  
    udp_ratio = grouped.apply(lambda x: (x['Protocol'] == 17).sum() / len(x), include_groups=False)
    #SYN flag count divided by total packets  
    tcp_syn_ratio = grouped.apply(lambda x: x['SYN Flag Count'].sum() / max(packets_per_window[x.name], 1), include_groups=False)
    #How many unique source IPs are in each window  
    unique_src_ip = grouped['Source IP'].nunique()
    #How many unique destination ports
    unique_dst_port = grouped['Destination Port'].nunique()
    #  
    flow_count = grouped.size()

    #If every packet in the window is benign label 0, if attack label 1
    label = grouped['Label'].apply(lambda x: 0 if (x == 'BENIGN').all() else 1, include_groups=False)

    #Create a dataframe to display our features
    windows_df = pd.DataFrame({
        'packets_per_window': packets_per_window,
        'bytes_per_window': bytes_per_window,
        'udp_ratio': udp_ratio,
        'tcp_syn_ratio': tcp_syn_ratio,
        'unique_src_ip': unique_src_ip,
        'unique_dst_port': unique_dst_port,
        'flow_count': flow_count,
        'label': label
    })

    #Places features into a single dataframe
    return windows_df.reset_index()

#Run the create_windows function with each window size
windows_1s = create_windows(combined_df, 1)
windows_5s = create_windows(combined_df, 5)
windows_10s = create_windows(combined_df, 10)

print("1s windows:", windows_1s['label'].value_counts().to_dict())
print("5s windows:", windows_5s['label'].value_counts().to_dict())
print("10s windows:", windows_10s['label'].value_counts().to_dict())
    
    
```
